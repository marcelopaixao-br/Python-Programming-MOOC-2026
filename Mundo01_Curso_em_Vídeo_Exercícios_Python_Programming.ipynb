````python
### Mini Aula: Operadores Aritméticos

Os operadores aritméticos são os blocos de construção da matemática no código.

1.

• Soma (+): Junta dois valores.
	• Exemplo: 5 + 3 resulta em 8.
• Subtração (-): Tira um valor do outro.
	• Exemplo: 10 - 4 resulta em 6.
• Multiplicação (*): Multiplica. Atenção: no computador usamos o asterisco, não o "x".
	• Exemplo: 4 * 3 resulta em 12.
• Divisão (/): Divide um valor pelo outro.
	• Exemplo: 10 / 2 resulta em 5.

2. Os Operadores Especiais

Aqui é onde a programação fica interessante. Estes três operadores costumam confundir quem está começando, mas são extremamente úteis:

Módulo ou Resto da Divisão (%)

Não calcula porcentagem! Ele divide dois números e devolve apenas o que sobrou dessa divisão.
• Como pensar: Se você tem 5 pedaços de pizza e divide igualmente entre 2 pessoas, cada uma ganha 2 pedaços inteiros e sobra 1.
• No código: 5 % 2 resulta em 1.
• Uso real: Muito usado para descobrir se um número é par ou ímpar (se numero % 2 der 0, é par; se der 1, é ímpar).

Exponenciação (**)

Eleva um número à potência de outro (base e expoente). Em algumas linguagens antigas usa-se o circunflexo (^), mas na maioria das modernas (como Python e JavaScript) usamos dois asteriscos.
• Exemplo: 2 ** 3 (2 ao cubo: 2 * 2 * 2) resulta em 8.

Divisão Inteira (//)

Divide dois números, mas ignora a parte quebrada (depois da vírgula), arredondando sempre para baixo.
• Exemplo: Numa divisão normal, 9 / 4 dá 2.25. Na divisão inteira, 9 // 4 resulta em 2.

3. A Regra de Ouro: Ordem de Precedência 🏆

O computador não lê a conta da esquerda para a direita cegamente. Ele segue uma ordem de prioridade estrita (regrinha de matemática):
1. Parênteses (): Tudo que estiver aqui dentro é feito primeiro.
2. Exponenciação **
3. Multiplicação *, Divisão /, Divisão Inteira // e Módulo % (Quem aparecer primeiro da esquerda para a direita).
4. Soma + e Subtração -
Veja a diferença:
• 5 + 3 * 2 ➡️ O computador faz 3 * 2 primeiro (= 6) e depois soma 5. Resultado: 11.
• (5 + 3) * 2 ➡️ O computador faz os parênteses primeiro (= 8) e depois multiplica por 2. Resultado: 16. ###
````

````python
# Exercício 007 - 02: Desenvolva um programa que leia as duas notas de um aluno, calcule e mostre a sua média.
nota1 = float(input("Digite a primeira nota: ").replace(',', '.'))
nota2 = float(input("Digite a segunda nota: ").replace(',', '.'))

total = nota1 + nota2
media = total / 2

print(f"O aluno tirou {nota1} e {nota2}, totalizando {total} e sua média foi {media:.1f}.")
````

````python
### Observação Importante: Números Decimais e o Método `.replace()`

Por padrão, o Python segue a notação matemática estadounidense:
* Usa **ponto (`.`)** para separar casas decimais (ex: `14.5`).
* Não utiliza pontos para separar milhares (ex: `1425.25`).

Ao tentar converter um número digitado com vírgula (ex: `"14,5"`) usando `float()`, o Python exibirá um erro (`ValueError`).

#### Como resolver na entrada de dados (`input`)

1. **Para números decimais simples (ex: `14,5`):**
   Use `.replace(',', '.')` para trocar a vírgula por ponto antes de converter:
   ```python
   valor = float(input("Digite um número: ").replace(',', '.'))
   ```

2. **Para números no padrão brasileiro completo com milhar e vírgula (ex: `1.425,25`):**
   Encadeie dois `.replace()` para primeiro remover o ponto de milhar e depois trocar a vírgula por ponto:
   ```python
   valor = float(input("Digite um valor: ").replace('.', '').replace(',', '.'))
   ```
````

````python
# Exercício 009: Faça um programa que leia um número inteiro qualquer e mostre na tela a sua tabuada.
numero = int(input("Digite um número: "))

print("-" * 12)
print(f"{numero} x {1:2} = {numero*1}")
print(f"{numero} x {2:2} = {numero*2}")
print(f"{numero} x {3:2} = {numero*3}")
print(f"{numero} x {4:2} = {numero*4}")
print(f"{numero} x {5:2} = {numero*5}")
print(f"{numero} x {6:2} = {numero*6}")
print(f"{numero} x {7:2} = {numero*7}")
print(f"{numero} x {8:2} = {numero*8}")
print(f"{numero} x {9:2} = {numero*9}")
print(f"{numero} x {10:2} = {numero*10}")
print("-" * 12)

````

````python
# Exercício 009 - 2: Faça um programa que leia um número inteiro qualquer e mostre na tela a sua tabuada.

numero = int(input("Digite um número: "))

print("-" * 12)
for i in range(1, 11):
    print(f"{numero} x {i:2} = {numero * i}")
print("-" * 12)
````
